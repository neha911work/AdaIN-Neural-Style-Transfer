```python
import os
import random
import shutil

painter_path = "/kaggle/input/datasets/kovalevvyu/painter-by-numbers-resized"
style_path = "/kaggle/working/NST_PROJECT/style_data"

os.makedirs(style_path, exist_ok=True)

# Get Painter images
images = [
    os.path.join(painter_path, f)
    for f in os.listdir(painter_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

print("Total Painter images:", len(images))

# Select 5,000 images
random.seed(42)
selected = random.sample(images, 5000)

# Copy them into the project's style_data folder
for i, src in enumerate(selected):
    destination = os.path.join(style_path, f"style_{i:05d}.jpg")
    shutil.copy2(src, destination)

print("Style images copied:", len(os.listdir(style_path)))
print("Style folder:", style_path)
```

```python
from PIL import Image, ImageOps, ImageDraw
from IPython.display import display

epochs = [5, 10, 15, 20]

images = []

for epoch in epochs:
    path = f"/kaggle/working/NST_PROJECT/experiment/kaggle_full_training/output_{epoch}.png"
    img = Image.open(path).convert("RGB")
    img = img.resize((400, 300))
    
    # Add epoch label
    canvas = Image.new("RGB", (400, 330), "white")
    canvas.paste(img, (0, 30))
    
    draw = ImageDraw.Draw(canvas)
    draw.text((10, 8), f"Epoch {epoch}", fill="black")
    
    images.append(canvas)

grid = Image.new("RGB", (800, 660), "white")

for i, img in enumerate(images):
    x = (i % 2) * 400
    y = (i // 2) * 330
    grid.paste(img, (x, y))

display(grid)
```

```python
from pathlib import Path

train_file = Path("/kaggle/working/train.py")
text = train_file.read_text()

old = """        if (epoch+1) % args.save_interval == 0:
                    torch.save(decoder.state_dict(), save_dir / f'decoder {epoch+1}.pth')
                    torch.save(optimizer.state_dict(), save_dir / f'optimizer {epoch+1}.pth')

                    with torch.no_grad():
                         output = torch.cat([content_batch, style_batch, g], dim=0)
                         save_image(output, save_dir / f'output_{epoch+1}.png', nrow=args.batch_size)
"""

new = """        if (epoch+1) % args.save_interval == 0:
            # Save decoder checkpoint
            torch.save(
                decoder.state_dict(),
                save_dir / f'decoder_{epoch+1}.pth'
            )

            # Save optimizer checkpoint
            torch.save(
                optimizer.state_dict(),
                save_dir / f'optimizer_{epoch+1}.pth'
            )

            # Save latest decoder separately
            torch.save(
                decoder.state_dict(),
                save_dir / 'decoder.pth'
            )

            # Save sample output
            with torch.no_grad():
                output = torch.cat([content_batch, style_batch, g], dim=0)
                save_image(
                    output,
                    save_dir / f'output_{epoch+1}.png',
                    nrow=args.batch_size
                )
"""

if old in text:
    text = text.replace(old, new)
    train_file.write_text(text)
    print("✅ Saving section fixed successfully!")
else:
    print("❌ The exact section was not found.")
```
